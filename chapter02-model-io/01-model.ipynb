```python
from langchain_openai import AzureChatOpenAI
from dotenv import load_dotenv

load_dotenv()

llm = AzureChatOpenAI(azure_deployment="gpt-5.5")

# 发送请求
print("🔄 正在请求 AI...")
response = llm.invoke("請說明MES系統?")

# 打印结果
print("\n🤖 AI 回答:")
print("=" * 60)
print(response.content)
print("=" * 60)
```
